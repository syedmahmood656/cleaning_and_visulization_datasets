#### plotting graphs to findout the relation between input columns and output column

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
sns.color_palette('Set2')

plt.figure(figsize=(6,6))
ax =sns.countplot(x='sex',data=data, order = ['Male','Female'],hue='income')
ax.set_xticklabels(['Male','Female'],rotation=40,ha="right")
plt.title("By Gender Distribution")
total= len(data)


for p in ax.patches:
    percentage = f'{100 * p.get_height() / total:.1f}%\n'
    x = p.get_x() + p.get_width() / 2
    y = p.get_height()
    ax.annotate(percentage, (x, y), ha='center', va='center')
```

from the above bar chart we can see that more than 45% of males are earning more than 50k, and 29% of females re reaning more than 50k, 20% of male and 3% of females are earning less than 50k
