```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('FlavorSense.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.isnull().mean()*100
```

```python
df['age'].unique()
```

```python
#handling missing values in age column by replacing it with mean of age
df['age'].fillna('43',inplace=True)
```

```python
df['age'].isnull().sum()
```

```python
#historical_cuisine_exposure
df['historical_cuisine_exposure'].value_counts()
```

```python
df['sleep_cycle'].value_counts()
```

```python
sc = df['sleep_cycle'].mode()[0]
df['sleep_cycle'].fillna(sc, inplace=True)
```

```python
mode = df['exercise_habits'].mode()[0]
df['exercise_habits'].fillna(mode, inplace=True)
```

```python
mode = df['climate_zone'].mode()[0]
df['climate_zone'].fillna(mode, inplace=True)
```

```python
mode = df['historical_cuisine_exposure'].mode()[0]
df['historical_cuisine_exposure'].fillna(mode, inplace=True)
```

```python
df.isnull().sum()
```
