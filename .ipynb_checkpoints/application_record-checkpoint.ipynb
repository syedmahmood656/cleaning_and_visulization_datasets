### visualization

```python
sns.countplot(x="NAME_EDUCATION_TYPE", data=df)
```

```python
sns.countplot(x='NAME_HOUSING_TYPE',data=df)
```

```python
sns.barplot(x='CNT_CHILDREN', y='DAYS_EMPLOYED', data=df)
```

```python
plt.scatter(df['AMT_INCOME_TOTAL'],df['DAYS_EMPLOYED'],color='red',marker='*')
```

```python
sns.scatterplot(data=df, x="DAYS_BIRTH", y="CNT_FAM_MEMBERS", hue ="FLAG_PHONE")
```

```python
sns.relplot(data=df, x="CNT_CHILDREN", y="CNT_FAM_MEMBERS", kind='line')
```

```python
plt.hist(df['AMT_INCOME_TOTAL'],bins=[2.610000e+04,3.610000e+04,4.610000e+04,5.610000e+04,6.610000e+04,7.610000e+04])
```

```python
y = df['OCCUPATION_TYPE'].value_counts()
plt.figure(figsize=(8, 6))
plt.pie(y, labels=y.index, autopct='%1.1f%%', pctdistance=0.85, colors=sns.color_palette('Set2'))

plt.show()
```

```python
y = df['NAME_HOUSING_TYPE'].value_counts()
plt.figure(figsize=(8, 6))
plt.pie(y, labels=y.index, autopct='%1.1f%%', pctdistance=0.85, colors=sns.color_palette('Set2'))

plt.show()
```
