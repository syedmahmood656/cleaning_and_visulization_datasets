### visualization

```python
sns.countplot(x='CATEGORY_NAME', data=df)
```

```python
sns.countplot(x='COUNTRY_NAME',data=df)
```

```python
sns.barplot(x='STATE', y='QUANTITY', data=df)
```

```python
sns.barplot(x='Gender', y='Weight', data=df)
```

```python
plt.scatter(df['STANDARD_COST'],df['LIST_PRICE'],color='red',marker='*')
```

```python
sns.scatterplot(data=df, x="STANDARD_COST", y="QUANTITY")
```

```python
sns.scatterplot(data=df, x="STANDARD_COST", y="QUANTITY",hue = "CATEGORY_NAME")
```

```python
sns.relplot(data=df, x="STANDARD_COST", y="STANDARD_COST", kind='line')
```

```python
plt.hist(df['STANDARD_COST'])
```

```python
y = df['CATEGORY_NAME'].value_counts()
plt.figure(figsize=(8, 6))
plt.pie(y, labels=y.index, autopct='%1.1f%%', pctdistance=0.85, colors=sns.color_palette('Set2'))

plt.show()
```
