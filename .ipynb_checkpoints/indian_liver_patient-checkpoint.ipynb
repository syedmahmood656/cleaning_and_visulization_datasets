### visualization

```python
sns.countplot(x="Gender", data=df)
```

```python
sns.barplot(x='Total_Protiens', y='Albumin', data=df)
```

```python
plt.scatter(df['Total_Protiens'],df['Albumin'],color='red',marker='*')
```

```python
plt.scatter(df['Direct_Bilirubin'],df['Total_Bilirubin'],color='red',marker='*')
```

```python
sns.residplot(data=df,x='Alkaline_Phosphotase',y='Alamine_Aminotransferase',lowess=True,line_kws=dict(color="r"))
```

```python
sns.scatterplot(data=df, x="Aspartate_Aminotransferase", y="Total_Protiens", hue ="Gender")
```

```python
sns.relplot(data=df, x="Age", y="Direct_Bilirubin", kind='line')
```
