### visualization

```python
sns.countplot(x="dominant_land_cover", data=df)
```

```python
sns.countplot(x='region',data=df)
```

```python
sns.barplot(x='dominant_land_cover', y='elevation', data=df)
```

```python
plt.scatter(df['tmp'],df['wet'],color='red',marker='*')
```

```python
sns.scatterplot(data=df, x="X5_Ct_2010_Da", y="X5_Bf_2010_Da", hue ="lumpy")
```

```python
sns.residplot(data=df,x='tmx',y='vap',lowess=True,line_kws=dict(color="r"))
```

```python
sns.relplot(data=df, x="elevation", y="wet", kind='line')
```

```python
plt.hist(df['dominant_land_cover'],bins=[0,2,4,6,8,10,12])
```
