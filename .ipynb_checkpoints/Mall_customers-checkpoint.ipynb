```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import seaborn.objects as so
```

```python
df = pd.read_csv("Mall_Customers.csv")
```

```python
df.head()
```

```python
(
    so.Plot(df, x="Gender", y="Annual Income (k$)")
    .add(so.Dots(pointsize=6))
    .add(so.Range(linewidth=2))
)
```

```python
so.Plot(df, x="Annual Income (k$)", color="Gender").add(so.Bar(), so.Hist(), so.Dodge())
```

```python
sns.relplot(data=df, x="Annual Income (k$)", y="Spending Score (1-100)", hue="Gender")
```

```python
sns.scatterplot(data=df, x="Age", y="Annual Income (k$)", hue="Gender")
```

```python
#may_flights = flights.query("month == 'May'")
sns.lineplot(data=df, x="Age", y="Spending Score (1-100)")
```

```python
sns.lineplot(data=df,x="Age",y="Annual Income (k$)")
```
