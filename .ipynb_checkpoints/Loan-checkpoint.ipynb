```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import numpy as np
```

```python
import seaborn.objects as so
```

```python
df = pd.read_csv("loan_approval.csv")
```

```python
df.sample(5)
```

```python
df.info()
```

```python
df['city'].nunique()
```

```python
df['years_employed'].nunique()
```

```python
p = so.Plot(df, "income", "loan_amount", color="loan_approved")
(
    p
    .add(so.Dash(), so.Agg(), so.Dodge())
    .add(so.Dots(), so.Dodge(), so.Jitter())
)
```

```python
p = so.Plot(df, "income", "loan_amount", color="loan_approved")
p.add(so.Dash(), so.Dodge())
```

```python
(
    df
    .rename_axis(index="credit_score")
    .pipe(so.Plot, x="loan_amount", ymin="years_employed", ymax="credit_score")
    .add(so.Range(), color="loan_approved")
)
```

```python
sns.jointplot(data=df, x="income", y="loan_amount", hue="loan_approved", kind="hist")
```

```python
so.Plot(df, x="loan_amount", color="loan_approved").add(so.Bar(), so.Hist(), so.Dodge())
```

```python
p = so.Plot(df, "points").scale(x="log")
p.add(so.Bars(), so.Hist(), so.Stack(), color="loan_approved")
```

```python
(
    so.Plot(df, "credit_score", "loan_amount", color="loan_approved")
    .add(so.Area(alpha=.7), so.Stack())
)
```

```python
so.Plot(df, "years_employed").add(so.Bar(), so.Hist())
```

```python
p = so.Plot(df, x="points")
p.add(so.Area(), so.KDE())
```

```python
p.add(so.Area(), so.KDE(), color="loan_approved")
```

```python
sns.scatterplot(data=df, x="income", y="credit_score", hue="loan_approved")
```
